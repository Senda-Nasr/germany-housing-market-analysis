**Exploring Categorical Column Values**

***Checking for Cardinality***

```python
#displaying number of unique values and missing percentage for each categorical column

cat_df = df.select_dtypes(exclude="number")
summary = pd.DataFrame({
    "unique_values": cat_df.nunique(),
    "missing_percentage": cat_df.isnull().mean() * 100
})

summary["missing_percentage"] = (
    summary["missing_percentage"].map(lambda x: f"{x:.2f}"))

#categorizing columns into low, medium, and high cardinality
def cardinality_label(x):
    if x <= 10:
        return "low"
    elif x <= 50:
        return "medium"
    else:
        return "high"

summary["cardinality"] = summary["unique_values"].apply(cardinality_label)
summary = summary.sort_values("unique_values", ascending=False)
summary
```

***Ordering by Percentage of Missing Values***

```python
summary['missing_percentage']=pd.to_numeric(summary['missing_percentage'])
summary = summary.sort_values("missing_percentage", ascending=False)
cat_cols_missing = summary[summary["missing_percentage"]>0]
display(cat_cols_missing)
```

***Inspecting lower cardinality categorical columns with missing values***

```python
low_cardinality_columns = cat_cols_missing[cat_cols_missing["unique_values"] <= 15].index

for col in low_cardinality_columns:
    print(f"\n{col}:")
    display(df[col].value_counts(dropna=False))
```
