**🧠 Concept you need for 8a — comparing two columns (a "boolean mask")**

Comparing two Series with `==` compares them **row by row** and returns a **new Series of `True`/`False`** — one entry per row. This is called a *mask* (or boolean Series). Nothing is filtered or summarized yet; you just get the mask.

- Alignment is by **index**, not position: row 0 is compared with row 0, row 5 with row 5. That is what we want here — same row = same listing.
- `True`/`False` behave like `1`/`0`, so a mask can be **summed** or **averaged**.
- `!=` builds the opposite mask directly (the *disagreements*); `~mask` flips any mask.

**Two-step recipe**

1. **Build the mask** — put your two columns on either side of `==` (or `!=`).
2. **Summarize the mask** — never try to eyeball 268,850 booleans:

| You want to know | Method on the mask |
|---|---|
| Do **all** rows agree? | `.all()` → a single `True`/`False` |
| **How many** rows differ? | sum the `!=` mask |
| **What fraction** differ? | `.mean()` of the mask (0.0–1.0) |

> 💡 Bonus, worth 30 seconds: `Series.equals()` answers *"are these two Series the same?"* in one call. It is **stricter** than `==` — it also checks dtype and index — and, unlike `==`, it treats a `NaN` in the same position as equal.
